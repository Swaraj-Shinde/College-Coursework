```python
import numpy as np
```

```python
# 1.Create a 1D NumPy array with values from 10 to 49
arr = np.arange(10, 50)
print("Array:",arr)
```

```python
# 2.Reshape the array into a 3x5 matrix
matrix = arr[:15].reshape(3, 5)
print("Matrix:",matrix)
```

```python
# 3.Extract elements divisible by 3
div_3 = arr[arr % 3 == 0]
print("\nElements divisible by 3:",div_3)
```

```python
# 4. Create two 3x3 random integer arrays
A = np.random.randint(1, 10, (3, 3))
B = np.random.randint(1, 10, (3, 3))
print("\narray a: ",A)
print("\narray b: ",B)
```

```python
# 5. Element-wise operations
print("\nAddition:",A+B)
print("\nSubtraction: ",A-B)
print("\nMultiplication: ",A*B)
print("\nDivision:",A/B)
```
