```python
#reverse a list
a=[10,20,30,40,50]
a[::-1]
```

```python
#remove duplicates from a list
a=[10,20,30,40,50,20,30]
unique=[]

for i in a:
    if i not in unique:
        unique.append(i)

print(unique)     
```

```python
#merge two dictionaries
d1={"name":"swaraj","age":20}
d2={"name1":"shinde","age1":21}
d1.update(d2)
print(d1)
```

```python
#sort the dictionary using key or values
d1={"c":30,"a":10,"b":20}
sorted_dict=dict(sorted(d1.items()))
print(sorted_dict)
```

```python
#intersection of two sets
a={10,20,30,40}
b={30,40,50,60}
result=a&b #a.intersection(b)
print(result)
print(a)
```

```python
#union of two sets
a={10,20,30,40}
b={30,40,50,60}
result=a.union(b) #a|b
print(result)
```
