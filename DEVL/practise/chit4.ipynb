Array Manipulations and Boolean Indexing
1. Create a 1D NumPy array with 15 evenly spaced numbers between 1 and 100.
2. Create a 4x4 array with random integers between 1 and 50 and replace all even numbers with -1.
3. Create a 5x5 matrix of random integers between 1 and 100 and retrieve all elements in second row
and last column
4. Create a NumPy array with integers from 1 to 50 and extract all the odd numbers.
5. Create a 3x3 array with random floating-point values and Calculate its mean

```python
import numpy as np
#create 1d array with evenly spaced numbers
arr=np.linspace(1,100,15)
print(arr)
```

```python
#create a 4x4 array and replace even numbers with -1
arr2=np.random.randint(1,51,16)
print("Before changing: \n",arr2.reshape(4,4))

arr2[arr2%2==0]=-1
print("After changing: \n",arr2.reshape(4,4))
```

```python
matrix=np.random.randint(1,101,(5,5))
print(matrix)

#retrieve 2nd row and last column
print("second row: ",matrix[1,:])
print("last column: ",matrix[:,4])
print("complete matrix: ",matrix[:,:])
```

```python
#array with int 1,50 and extract all odd numbers
a=np.random.randint(1,51,10)
print(a)

b=a[a%2!=0]
print(b)
```

```python
#random floating point 3x3 array and calculate its mean
a=np.random.rand(3,3)
print(a) 
print("Mean: ",a.mean())
```
