```python
import pandas as pd
import requests
from io import StringIO

url = "https://en.wikipedia.org/wiki/World_population"
#url variable stores address of webpage to be scraped

headers = {"User-Agent": "Mozilla/5.0"}
#sometimes access requests get denied from program
#this line tells page to treat me like a normal browser
#who wants to access the webpage
#dictionary

response = requests.get(url, headers=headers)
response.raise_for_status()

tables = pd.read_html(StringIO(response.text))

print("Total tables:", len(tables))

for i, table in enumerate(tables):
    print(f"\nTable {i+1}")
    print(table.head())
    table.to_csv(f"table_{i+1}.csv", index=False)
```

```python
import requests
from bs4 import BeautifulSoup

url = "https://en.wikipedia.org/wiki/World_population"

headers = {
    "User-Agent": "Mozilla/5.0"
}

response = requests.get(url, headers=headers)

soup = BeautifulSoup(response.text, "html.parser")

paragraphs = soup.find_all("p")

for i, p in enumerate(paragraphs, 1):
    text = p.get_text(strip=True)
    if text:
        print(f"Paragraph {i}: {text}\n")
```

```python
import requests
from bs4 import BeautifulSoup

url = "https://en.wikipedia.org/wiki/World_population"

headers = {
    "User-Agent": "Mozilla/5.0"
}

response = requests.get(url, headers=headers)
soup = BeautifulSoup(response.text, "html.parser")

images = soup.find_all("img")

print("Total Images:", len(images))

for img in images:
    print(img.get("src"))
```

```python
import requests
from bs4 import BeautifulSoup

url = "https://en.wikipedia.org/wiki/World_population"

response = requests.get(url, headers={"User-Agent": "Mozilla/5.0"})
soup = BeautifulSoup(response.text, "html.parser")

for link in soup.find_all("a"):
    href = link.get("href")
    if href:
        print(href)
```
