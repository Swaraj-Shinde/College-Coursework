```python
import sqlite3
conn = sqlite3.connect("student.db")
cursor = conn.cursor()
cursor.execute("""
CREATE TABLE IF NOT EXISTS Students2 (
    ID INTEGER,
    Name TEXT,
    Age INTEGER,
    Marks INTEGER
)
""")
cursor.execute("INSERT INTO Students2 VALUES (1, 'Bhavik', 20, 85)")
cursor.execute("INSERT INTO Students2 VALUES (2, 'Bharat', 21, 90)")
cursor.execute("INSERT INTO Students2 VALUES (3, 'Arvind', 22, 88)")
cursor.execute("INSERT INTO Students2 VALUES (4, 'Amit', 20, 92)")
conn.commit()
```

```python
df = pd.read_sql_query("SELECT * FROM Students2",conn)
print("Head: ")
print(df.head())
```

```python
print(df.tail())
print(df.info())
print(df.describe())
print(df.dtypes)
print(df.shape)
print(df.columns)
conn.close()
```
