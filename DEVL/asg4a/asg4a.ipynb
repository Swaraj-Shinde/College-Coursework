```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from scipy import stats
#scipy.stats contains many tests like t test,chi test,anova


sample1 = [23, 45, 31, 50, 41, 47]
sample2 = [22, 31, 29, 34, 42]

# T-Test
print("One-Sample T-Test p-val", stats.ttest_1samp(sample1, 40).pvalue)
print("Two-Sample T-Test p-val", stats.ttest_ind(sample1, sample2).pvalue)


# Chi Test
contingency_table = [[10, 20], [30, 40]]
chi2, p_val, dof, expected = stats.chi2_contingency(contingency_table)
print("Chi-Square Test p-value:", p_val)

# ANOVA
g1, g2, g3 = [20, 30, 40], [25, 35, 45], [22, 32, 42]
print("ANOVA p-value:", stats.f_oneway(g1, g2, g3).pvalue)
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
data = np.random.normal(loc=0, scale=1, size=1000)
sns.histplot(data, kde=True)
plt.title('Normal Distribution')
plt.show()
```

```python
data_binom = np.random.binomial(n=10, p=0.5, size=1000)
sns.histplot(data_binom, kde=False)
plt.title('Binomial Distribution')
plt.show()
```

```python
data_poisson = np.random.poisson(lam=3, size=1000)
sns.histplot(data_poisson, kde=False)
plt.title('Poisson Distribution')
plt.show()
```
